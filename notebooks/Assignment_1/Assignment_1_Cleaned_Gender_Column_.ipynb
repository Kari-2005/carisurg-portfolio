```python
# Import the libraries needed for data cleaning
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries are successfully imported for data cleaning!")
```

```python
# Load the dataset
FILE_PATH = "/content/drive/MyDrive/CARISURG/Week0/EmergencyTriageDataset_Reduced_Dirty.csv"

df_raw = pd.read_csv(FILE_PATH)
print(f"Dataset Loaded: {df_raw.shape[0]} rows x {df_raw.shape[1]} columns")
print(f"Columns: {list(df_raw.columns)}")
```

```python
# Examining Rows
df_raw.head(10)
```

```python
# Looking at what types panda thinks each column is
print(df_raw.dtypes)
```

```python
# Explore the values in the Gender Column
print("Unique values in Gender:")
print(df_raw['Gender'].unique())
print(f"\nTotal unique values: {df_raw['Gender'].nunique()}")
```

```python
# Quantity of each unique value stored in Gender Column
print("Gender value quantity:")
print(df_raw['Gender'].value_counts())
```

```python
#In this design, gender is converted into numerical values:
# 1 - Male, 0 = Female. (Non-binary = 2 for future consideration)
# The current dataset only contains Male/ Female values but non-binary values will be included for future handling of genders.
# For a more efficient/ alternate way we can first convert all values to lowercase

gender_map = {
    'male': 1,
    'female': 0,
    '1': 1,
    '0': 0,
    # adding future considerations (m/f/non-binary)
    'm': 1,
    'f': 0,
    'non-binary': 2,
    'nonbinary': 2,
    'nb': 2,
    '2': 2
}
```

```python
# Cleaning the column (for efficient mapping we convert each to lowercase before we map
df_raw['Gender_Cleaned'] = (df_raw['Gender'].astype(str).str.strip().str.lower().map(gender_map))
```

```python
# Verifying the result after cleaning
print("After Cleaning:")
print(df_raw['Gender_Cleaned'].value_counts())
print(f"\nAny NaN values? {df_raw['Gender_Cleaned'].isnull().sum()}")
```

```python
# Replacing the original dirty column with the new cleaned data
if 'Gender' in df_raw.columns: df_raw = df_raw.drop(columns=['Gender'])

df_raw = df_raw.rename(columns={'Gender_Cleaned': 'Gender'})

# Confirm the result of the new columns
print("Column 'Gender' after cleaning:")
print(df_raw['Gender'].value_counts())
print(f"Data type: {df_raw['Gender'].dtype}")

df_raw.head()
```
