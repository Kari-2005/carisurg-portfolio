 **MAP Cleaning — SBP and DBP:**

Before calculating Mean Arterial Pressure (MAP), I decided to clean the SBP and DBP columns first because MAP is directly calculated from these two values. SBP represents the systolic blood pressure, while DBP represents the diastolic blood pressure. If either of these columns contains missing values, text values, or unrealistic readings, the MAP result may become inaccurate.

Since MAP is calculated using the formula:

MAP = DBP + (SBP - DBP) / 3

both SBP and DBP need to be valid before applying the calculation. For example, if SBP or DBP has an out-of-range value, that error would carry over into the MAP column. Because of this, I cleaned the blood pressure values first by checking their data types, identifying missing values, and looking for values outside a reasonable medical range. This helped ensure that the final MAP values were based on reliable blood pressure readings.

```python
# SBP = Systolic Blood Pressure Cleaning.
```

```python
# Valid range: 50–250 mmHg.
# First we inspect

print("SBP unique values (sample):", df['SBP'].unique()[:15])
print("SBP dtype:", df['SBP'].dtype)
```

```python
# Convert to non-numerical entries to numerical (becomes NaN)
df['SBP'] = pd.to_numeric(df['SBP'], errors='coerce')
print("After Numerical conversion:")
print(df['SBP'].describe())
```

```python
# Identify any values that are out-of-range
invalid_sbp = df[(df['SBP'] < 50) | (df['SBP'] > 250)]
print(f"Out-of-range SBP rows: {len(invalid_sbp)}")
print(invalid_sbp[['ID', 'SBP']].head())
```

```python
# Replace any out-of-range values with NaN, then impute with median.
# The median was used to impute SBP because it is not strongly affected by outliers.
# This makes it a more reliable replacement value for missing or invalid blood pressure
# readings than the mean.

df.loc[(df['SBP'] < 50) | (df['SBP'] > 250), 'SBP'] = np.nan
sbp_median = df['SBP'].median()
df['SBP'] = df['SBP'].fillna(sbp_median)

print(f"SBP after cleaning: min={df['SBP'].min()}, max={df['SBP'].max()}, NaNs={df['SBP'].isnull().sum()}")
```

```python
# DSP = Diastolic Blood Pressure
```

```python
# Valid Range: 30-150 mmHg
# First we inspect

print("DBP dtype:", df["DBP"].dtype)
print("DBP unique values sample:", df["DBP"].unique()[:20])
print("DBP missing values before cleaning:", df["DBP"].isnull().sum())
```

```python
# Converting DBP to numeric
df["DBP"] = pd.to_numeric(df["DBP"], errors="coerce")

print("\nDBP dtype after conversion:", df["DBP"].dtype)
print("DBP missing values after conversion:", df["DBP"].isnull().sum())
```

```python
# Flagging invalid DBP values
# Valid DBP range: 30-150 mmHg
invalid_dbp = df[(df["DBP"] < 30) | (df["DBP"] > 150)]

print(f"\nDBP values outside valid range (30-150): {len(invalid_dbp)} rows")
print(invalid_dbp[["ID", "DBP"]].head(10))
```

```python
# Replacing invalid DBP values with NaN
df.loc[(df["DBP"] < 30) | (df["DBP"] > 150), "DBP"] = np.nan

print("\nDBP missing values after range filtering:", df["DBP"].isnull().sum())
```

```python
# Imputing missing DBP values using the median
# Median is used because blood pressure values can contain outliers,
# and the median is less affected by extreme values than the mean.

dbp_median = df["DBP"].median()

df["DBP"] = df["DBP"].fillna(dbp_median)

print(f"\nMedian DBP used for imputation: {dbp_median}")
print("DBP missing values after median imputation:", df["DBP"].isnull().sum())

print("\nDBP summary after imputation:")
print(df["DBP"].describe())

print("DBP missing values after range filtering:", df["DBP"].isnull().sum())
```

```python
# MAP- Mean Arterial Pressure
```

```python
# Valid Range: 40-180 mmHg
# Calculating MAP

df["MAP"] = (df["SBP"] + 2 * df["DBP"]) / 3
```

```python
# After calculating MAP, we inspect the new values

print("MAP missing values:", df["MAP"].isnull().sum())
print("MAP range:", df["MAP"].min(), "-", df["MAP"].max())
print(df[["SBP", "DBP", "MAP"]].head())
```

```python
# Check for invalid MAP values

invalid_map = df[(df["MAP"] < 40) | (df["MAP"] > 180)]

print("MAP values outside valid range:", len(invalid_map))
print(invalid_map[["SBP", "DBP", "MAP"]].head())
```

```python
# Flag MAP values that may be clinically low but still possible
# <60 and >180 is alarming as this means that there is a serious medical problem.

df["MAP_flag"] = "Normal/Acceptable"

df.loc[df["MAP"] < 60, "MAP_flag"] = "Low MAP - review"
df.loc[df["MAP"] > 180, "MAP_flag"] = "High MAP - review"

# Display rows where MAP needs review
print(df[df["MAP_flag"] != "Normal/Acceptable"][["SBP", "DBP", "MAP", "MAP_flag"]])
```

```python
# Final validation check for SBP, DBP, and MAP

# A list of columns to check, along with their accepted minimum and maximum values
checks = [
    ("SBP", 50, 250),    # SBP should be between 50 and 250 mmHg
    ("DBP", 30, 150),    # DBP should be between 30 and 150 mmHg
    ("MAP", 40, 180)     # MAP should be between 40 and 180 mmHg
]

# Validation results
print("Final Validation Check")
print("----------------------")

# Loop through each column and its valid range
for col, lo, hi in checks:

    # Count how many missing values are in the column
    n_missing = df[col].isnull().sum()

    # Count how many values are outside the accepted range
    n_invalid = ((df[col] < lo) | (df[col] > hi)).sum()

    # Find the minimum and maximum values in the column
    min_val = df[col].min()
    max_val = df[col].max()

    # If there are no missing or invalid values, the column passes the check
    if n_missing == 0 and n_invalid == 0:
        status = "PASS"
    else:
        status = "CHECK"

    # Print the validation results for the current column
    print(f"{status} {col}:")
    print(f"  Range: {min_val:.1f} - {max_val:.1f}")
    print(f"  Missing values: {n_missing}")
    print(f"  Out-of-range values: {n_invalid}")
    print()
```

```python
# Display final cleaned blood pressure columns

print(df[["SBP", "DBP", "MAP"]].head(15))
```
